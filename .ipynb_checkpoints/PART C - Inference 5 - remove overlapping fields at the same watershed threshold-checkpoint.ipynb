```python
data = pd.read_csv('inference_data/inference_data.csv')
tile_intersection_data = pd.read_csv('inference_data/tile_intersections.csv')
tile_intersection_data['tile_id'] = tile_intersection_data['tile_id'].apply(lambda x: f"{x:05d}")
tile_intersection_data['intersecting_tile_ids'] = tile_intersection_data['intersecting_tile_ids'].apply(lambda x: x[1:-1].split(", "))
image_filenames = data[data["num_cultivated_pixels"] > 0]["image_filename"].tolist() # images that have at least one cultivated pixel
tile_ids = [image_filename[-9:-4] for image_filename in image_filenames]
```

```python
def remove_overlapping_fields(start, stop):
    for i in tqdm(range(start, stop), smoothing=0):
        tile_id = tile_ids[i]
        tiles_already_saved = tile_ids[:i]
        tiles_to_check = tile_intersection_data[tile_intersection_data['tile_id']==tile_id]['intersecting_tile_ids'].iloc[0]
        tiles_to_check.remove(tile_id)
        tiles_to_check = list(set(tiles_to_check).intersection(set(tiles_already_saved)))

        df_a = pd.read_csv(f"inference_data/outputs/field_predictions/{tile_id}.csv")
        df_a['geometry'] = df_a['geometry'].apply(lambda x: loads(x))
        gdf_a = gpd.GeoDataFrame(df_a, geometry=df_a['geometry'], crs="EPSG:5070")
        for tile_id_to_check in tiles_to_check:
            df_b = pd.read_csv(f"inference_data/outputs/field_predictions_final/{tile_id_to_check}.csv")
            df_b['geometry'] = df_b['geometry'].apply(lambda x: loads(x))
            gdf_b = gpd.GeoDataFrame(df_b, geometry=df_b['geometry'], crs="EPSG:5070")
            for ranking in [1, 2, 3, 4]:
                gdf_a_ranking = gdf_a[gdf_a['priority_ranking']==ranking]
                gdf_b_ranking = gdf_b[gdf_b['priority_ranking']==ranking]
                intersecting_indices = np.unique(gpd.sjoin(gdf_a_ranking, gdf_b_ranking, how="inner", predicate="intersects").index)
                gdf_a.drop(np.unique(intersecting_indices), inplace=True)
    
        gdf_a.to_csv(f'inference_data/outputs/field_predictions_final/{tile_id}.csv', index=False)
```
